```python
import pandas as pd
import numpy as np
df = pd.read_csv('../data/20240121_shopee_sample_data (1).csv')
pd.set_option('display.max_rows',None)

# checks all the countries that have products listed on shopee
# location_counts = df['specification'].str.extract(r'Ships From\s+(.*)$')[0].value_counts()
# print(location_counts)

overseas_countries = ['China','Indonesia','Korea','Vietnam','Taiwan','Thailand','ID','JAKARTA']

# extracts the raw origin location of every product
df['raw_location'] = df['specification'].str.extract(r'Ships From\s+(.*)$')

pattern = '|'.join(overseas_countries)

df['origin_type'] = np.where(df['raw_location'].str.contains(pattern,case=False,na=False),'Overseas','Malaysia')

# shows the number of products that are local and overseas
print(df['origin_type'].value_counts())
df.to_csv('../data/20240121_shopee_sample_data (1).csv',index=False)
df.head(50)
```
