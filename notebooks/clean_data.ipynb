```python
import pandas as pd
from datetime import datetime
import math
df = pd.read_csv('../data/20240121_shopee_sample_data (1).csv')

# convert '8.1k' format into integers
def convert_k_to_num(val):
    if isinstance(val, str):
        if val.lower().endswith('k'):
            val = val.strip('k')
            try:
                return int(float(val)*1000)
            except ValueError:
                return val
    return val

#convert timestamps into datetime
def convert_timestamp(timestamp):
    if timestamp:
        dt = datetime.fromtimestamp(timestamp/1000)
        return dt
    return timestamp

df = df.dropna(subset=['price_actual', 'specification', 'item_rating', 'seller_name', 'total_rating','total_sold'])

print(df.isna().sum()) 

df_blanks = df[df['seller_name'].isna()]
print(df_blanks)

# delivery is not needed as too many null values
# df = df.drop(columns=['delivery']) 

df[['total_rating','total_sold']] = df[['total_rating','total_sold']].map(convert_k_to_num)

# df['timestamp'] = df['timestamp'].map(convert_timestamp)

df['price_actual'] = df['price_actual'].fillna(df['price_ori'])

# transforming the item_category_detail column into separate main categories and sub categories
# category_splits = df['item_category_detail'].str.split(' \| ', expand=True, regex=True)
# df['main_category'] = category_splits[1]
# df['sub_category1'] = category_splits[2]
# df['sub_category2'] = category_splits[3]

# df = df.drop(columns='item_category_detail')

df.to_csv('../data/20240121_shopee_sample_data (1).csv',index=False)


df.head(50)
```
